## Exercise 1: Valid Sudoku Checker
Implementing a Sudoku validator using NumPy array operations. A valid Sudoku must satisfy:
1. Each row contains digits 1-9 exactly once (sum = 45)
2. Each column contains digits 1-9 exactly once (sum = 45)
3. Each 3x3 sub-grid contains digits 1-9 exactly once (sum = 45)

```python
# Creating a complete 9x9 Sudoku grid for validation
# This is a valid, solved Sudoku puzzle
# Each row, column, and 3x3 box should contain digits 1-9 exactly once
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],  # Row 1
    [6, 7, 2, 1, 9, 5, 3, 4, 8],  # Row 2
    [1, 9, 8, 3, 4, 2, 5, 6, 7],  # Row 3
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],  # Row 4
    [4, 2, 6, 8, 5, 3, 7, 9, 1],  # Row 5
    [7, 1, 3, 9, 2, 4, 8, 5, 6],  # Row 6
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],  # Row 7
    [2, 8, 7, 4, 1, 9, 6, 3, 5],  # Row 8
    [3, 4, 5, 2, 8, 6, 1, 7, 9]   # Row 9
])
```

```python
# Validating rows: Check if each row sums to 45
# np.sum(s, axis=1) calculates sum along axis 1 (across columns for each row)
# Since 1+2+3+4+5+6+7+8+9 = 45, each valid row should sum to 45
rows = np.sum(s, axis = 1)

# Check each row sum - if any row doesn't sum to 45, Sudoku is invalid
for i in rows:
    if i != 45:
        print("Sudoku is invalid")
        break  # Exit loop immediately if invalid row found
else:
    # This else clause executes only if the loop completes without breaking
    print("For rows Sudoku is valid")
```

```python
# Validating columns: Check if each column sums to 45
# np.sum(s, axis=0) calculates sum along axis 0 (down rows for each column)
# Each valid column should also sum to 45
column = np.sum(s, axis = 0)

# Check each column sum - if any column doesn't sum to 45, Sudoku is invalid
for i in column:
    if i != 45:
        print("Sudoku is invalid")
        break  # Exit loop immediately if invalid column found
else:
    # This else clause executes only if the loop completes without breaking
    print("For columns Sudoku is valid")
```

```python
# Validating 3x3 sub-grids: Check if each 3x3 block sums to 45
# There are 9 sub-grids in a 9x9 Sudoku, arranged in a 3x3 pattern
# We iterate through starting positions: (0,0), (0,3), (0,6), (3,0), etc.

for i in range(0, 9, 3):  # Row starting positions: 0, 3, 6
    for j in range(0, 9, 3):  # Column starting positions: 0, 3, 6
        # Extract 3x3 sub-grid using slicing: s[i:i+3, j:j+3]
        # This gets a 3x3 block starting at position (i,j)
        if np.sum(s[i:i+3, j:j+3]) != 45:
            print("Sudoku is invalid")
            break  # Break inner loop if invalid block found
    else:
        # Continue outer loop if inner loop completed without breaking
        continue
    # This break executes only if inner loop was broken
    break  # Break outer loop
else:
    # This executes only if both loops completed without breaking
    print("Sudoku is valid for all 3x3 blocks")
```

## Exercise 2: Student Data Analysis
Analyzing student performance data using NumPy operations. This demonstrates practical data manipulation techniques commonly used in data science.

```python
# Creating student dataset with 3 columns: Age, Math Marks, Science Marks
# Each row represents one student's data
# This simulates a real-world dataset that might come from a school database
data = np.array([
    [18, 85, 78],   # Student 1: 18 years old, 85 in Math, 78 in Science
    [19, 92, 88],   # Student 2: 19 years old, 92 in Math, 88 in Science
    [17, 76, 95],   # Student 3: 17 years old, 76 in Math, 95 in Science
    [18, 65, 70],   # Student 4: 18 years old, 65 in Math, 70 in Science
    [20, 90, 85]    # Student 5: 20 years old, 90 in Math, 85 in Science
])

print("Student Data Structure:")
print("Columns: [Age, Math Marks, Science Marks]")
print("Data:")
print(data)
```

### Question 1: Get the shape of the matrix

```python
# Get the shape of the matrix
# .shape returns a tuple (rows, columns)
# Result: (5, 3) means 5 students (rows) and 3 attributes (columns)
data.shape
```

### Question 2: Find the average age of students

```python
# Find the average age of students
# data[:,0] selects all rows (:) and column 0 (age column)
# np.mean() calculates the arithmetic mean
# Ages: [18, 19, 17, 18, 20] → Average: (18+19+17+18+20)/5 = 18.4
np.mean(data[:,0])
```

### Question 3: Extract Math marks of all students

```python
# Extract Math marks of all students
# data[:,1] selects all rows and column 1 (Math marks column)
# Note: data[:data.size,1] is unnecessarily complex - data[:,1] is simpler
# Result: [85, 92, 76, 65, 90] - Math marks for all 5 students
math_marks = data[:,1]  # Simplified from original data[:data.size,1]
math_marks
```

### Question 4: Find the highest Science mark

```python
# Find the highest Science mark
# data[:,2] selects all rows and column 2 (Science marks column)
# np.max() finds the maximum value in the array
# Science marks: [78, 88, 95, 70, 85] → Maximum: 95
highest_sci_marks = np.max(data[:,2])
highest_sci_marks
```

### Question 5: Get details of students who scored more than 90 in Math

```python
# Get details of students who scored more than 90 in Math
# data[:,1] > 90 creates a boolean mask for Math marks > 90
# data[boolean_mask] returns only rows where the condition is True
# Only Student 2 (19 years, 92 Math, 88 Science) scored > 90 in Math
data[data[:,1] > 90]
```

```python
# Display current data before modification
# This helps track changes made to the dataset
data
```

### Question 6: Increase Math marks of all students by 5

```python
# Increase Math marks of all students by 5
# data[:,1] += 5 adds 5 to each element in column 1 (Math marks)
# This modifies the original array in-place
# Before: [85, 92, 76, 65, 90] → After: [90, 97, 81, 70, 95]
data[:,1] += 5
data
```

### Question 7: Find how many students are younger than 19

```python
# Find how many students are younger than 19
# data[:,0] < 19 creates boolean mask for ages < 19
# data[boolean_mask] returns rows where condition is True
# len() counts the number of rows returned
# Ages: [18, 19, 17, 18, 20] → Students < 19: [18, 17, 18] → Count: 3
len(data[data[:,0] < 19])
```

```python
# Display current data state
# Shows data after Math marks were increased by 5
data
```

### Question 8: Calculate the average marks in each subject (column-wise mean)

```python
# Calculate the average marks in each subject (column-wise mean)
# data[:,1:] selects all rows and columns 1-2 (Math and Science marks)
# axis=0 calculates mean along rows (i.e., for each column)
# Result: [86.6, 83.2] → Math average: 86.6, Science average: 83.2
# Alternative: np.mean(data[:,1]), np.mean(data[:,2]) for individual subjects
np.mean(data[:,1:], axis = 0)
```

### Question 9: Get data of students who scored at least 80 in both subjects

```python
# Get data of students who scored at least 80 in both subjects
# (data[:,1] >= 80) creates boolean mask for Math marks >= 80
# (data[:,2] >= 80) creates boolean mask for Science marks >= 80
# & operator combines both conditions (logical AND)
# Students meeting both criteria: Student 2 (97,88), Student 3 (81,95), Student 5 (95,85)
data[(data[:,1] >= 80) & (data[:,2] >= 80)]
```

### Question 10: Replace all Science marks < 75 with 0

```python
# Replace all Science marks < 75 with 0
# data[:,2] selects Science marks column
# data[:,2] < 75 creates boolean mask for Science marks < 75
# data[:,2][boolean_mask] = 0 sets those values to 0
# Science marks: [78, 88, 95, 70, 85] → Only 70 < 75, so it becomes 0
data[:,2][data[:,2] < 75] = 0
```

```python
# Display final data after all modifications
# Notice Student 4's Science mark changed from 70 to 0
# All Math marks were increased by 5 earlier
data
```

## Additional Practice Exercises
More challenging problems to reinforce NumPy concepts

```python
# Bonus Exercise 1: Find students with Math marks above average
math_avg = np.mean(data[:,1])
above_avg_math = data[data[:,1] > math_avg]
print(f"Math average: {math_avg:.1f}")
print("Students with above-average Math marks:")
print(above_avg_math)

# Bonus Exercise 2: Calculate total marks for each student
total_marks = np.sum(data[:,1:], axis=1)  # Sum Math and Science marks
data_with_totals = np.column_stack([data, total_marks])
print("\nData with total marks (Age, Math, Science, Total):")
print(data_with_totals)

# Bonus Exercise 3: Find the student with highest total marks
best_student_idx = np.argmax(total_marks)
print(f"\nBest performing student (index {best_student_idx}):")
print(f"Age: {data[best_student_idx, 0]}, Math: {data[best_student_idx, 1]}, Science: {data[best_student_idx, 2]}")
print(f"Total marks: {total_marks[best_student_idx]}")

# Bonus Exercise 4: Create grade categories
def assign_grades(marks):
    """Assign letter grades based on marks"""
    grades = np.empty(marks.shape, dtype='U1')  # Unicode string array
    grades[marks >= 90] = 'A'
    grades[(marks >= 80) & (marks < 90)] = 'B'
    grades[(marks >= 70) & (marks < 80)] = 'C'
    grades[(marks >= 60) & (marks < 70)] = 'D'
    grades[marks < 60] = 'F'
    return grades

math_grades = assign_grades(data[:,1])
science_grades = assign_grades(data[:,2])

print("\nGrade Report:")
for i in range(len(data)):
    print(f"Student {i+1}: Math {data[i,1]} ({math_grades[i]}), Science {data[i,2]} ({science_grades[i]})")

# Bonus Exercise 5: Statistical analysis
print("\nStatistical Summary:")
print(f"Math - Mean: {np.mean(data[:,1]):.1f}, Std: {np.std(data[:,1]):.1f}, Range: {np.ptp(data[:,1])}")
print(f"Science - Mean: {np.mean(data[:,2]):.1f}, Std: {np.std(data[:,2]):.1f}, Range: {np.ptp(data[:,2])}")
print(f"Age - Mean: {np.mean(data[:,0]):.1f}, Median: {np.median(data[:,0]):.1f}")
```
