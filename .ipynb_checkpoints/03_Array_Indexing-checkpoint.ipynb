```python
import numpy as np
```

**NUMPY INDEXING AND SLICING OF VECTORS**

```python
arr = np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[1:5]
```

```python
arr[:5]
```

```python
arr[:]
```

```python
arr[3:]
```

```python
arr[3::]
```

```python
arr_slice = arr[3::2]
arr_slice
```

**NUMPY INDEXING AND SLICING OF MATRIX**

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[5]
```

```python
arr[0][3]
```

```python
arr[5,4]
```

```python
slice_arr = arr[0:2,1:3]
slice_arr
```

```python
slice_arr = arr[3:,3:]
slice_arr
```

```python
arr[:,3]
```

**BOOLEAN INDEXING**

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr%2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
