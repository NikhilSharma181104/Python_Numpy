*Matrix Operation*

```python
mat1 = np.arange(1,5).reshape(2,2)
mat2 = np.arange(5,9).reshape(2,2)
```

```python
mat1 + mat2
```

```python
mat1 - mat2
```

```python
mat1 @ mat2
```

```python
np.dot(mat1, mat2)
```

```python
mat1.T
```

**Advance Array Manipulation**

*Stacking Array*

```python
a = np.arange(1,5)
b = np.arange(5,9)
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```
