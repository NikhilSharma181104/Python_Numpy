## Introduction to NumPy
NumPy (Numerical Python) is the fundamental package for scientific computing in Python. It provides:
- Powerful N-dimensional array objects
- Sophisticated broadcasting functions
- Tools for integrating C/C++ and Fortran code
- Useful linear algebra, Fourier transform, and random number capabilities

```python
# Importing NumPy library with standard alias 'np'
# This is the conventional way to import NumPy
import numpy as np

# Creating our first NumPy array from a Python list
# np.array() converts Python list to NumPy array
b = np.array([1, 2, 3])

# Printing the array - notice no commas between elements
print(b)

# Checking the type - it's numpy.ndarray (n-dimensional array)
print(type(b))
```

## Comparison: Python Lists vs NumPy Arrays

```python
# Demonstrating the difference between Python lists and NumPy arrays

# Python list - can contain mixed data types
python_list = [1, 2.5, "hello", True]
print("Python list with mixed types:", python_list)
print("Type:", type(python_list))

# NumPy array - all elements must be the same type
# When mixed types are provided, NumPy converts them to a common type
numpy_array_mixed = np.array([1, 2.5, "hello"])
print("\nNumPy array from mixed types:", numpy_array_mixed)
print("Data type:", numpy_array_mixed.dtype)  # All converted to strings (Unicode)

# NumPy array with numbers only
numpy_array_numbers = np.array([1, 2, 3.5])
print("\nNumPy array with numbers:", numpy_array_numbers)
print("Data type:", numpy_array_numbers.dtype)  # All converted to floats
```

```python
# Performance comparison between lists and arrays
import time

# Create large datasets
size = 1000000
python_list = list(range(size))
numpy_array = np.arange(size)

# Time list comprehension
start_time = time.time()
list_result = [x * 2 for x in python_list]
list_time = time.time() - start_time

# Time NumPy vectorized operation
start_time = time.time()
array_result = numpy_array * 2
array_time = time.time() - start_time

print(f"List operation time: {list_time:.4f} seconds")
print(f"NumPy operation time: {array_time:.4f} seconds")
print(f"NumPy is {list_time/array_time:.1f}x faster!")
```

## Key Advantages of NumPy Arrays

1. **Speed**: NumPy operations are implemented in C, making them much faster
2. **Memory Efficiency**: Arrays use less memory than Python lists
3. **Vectorization**: Operations are applied to entire arrays without explicit loops
4. **Broadcasting**: Operations between arrays of different shapes
5. **Mathematical Functions**: Extensive library of mathematical functions
