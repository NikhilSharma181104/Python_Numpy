**Array Generation Function**

```python
arr = np.arange(1,11,2)
print(arr)
```

```python
arr = np.zeros(6)
print(arr)
```

```python
arr = np.zeros((6,6))
print(arr)
```

```python
arr = np.ones(6)
print(arr)
```

```python
arr = np.ones((4,8))
arr
```

```python
arr = np.linspace(1,6,2)
arr
```

```python
arr = np.linspace(1,6,3)
arr
```

```python
arr = np.linspace(1,6,6)
arr
```

```python
arr = np.linspace(1,6,10)
arr
```

```python
arr = np.linspace(0,1,100)
arr
```
