### 1. Conditional Logic & Feature Creation: `np.where()`
`np.where(condition, [x, y])` functions as a vectorized ternary operator (`if condition then x else y`).
- **If condition, x, and y are provided**: Returns elements chosen from `x` or `y` depending on `condition`.
- **If only condition is provided**: Returns a tuple of arrays containing the indices where the condition evaluates to `True`.
- **Analytics Use Cases**: Creating binary classification flags (Pass/Fail, High/Low), applying tiered bonuses, and conditional imputation without mutating the original data.


```python
# 1. Basic ternary replacement (if-else logic)
scores = np.array([45, 82, 91, 58, 73, 33, 95])
status = np.where(scores >= 60, "Pass", "Fail")
print("Original Scores:", scores)
print("Status (Pass / Fail):", status)

# 2. Conditional data transformation
# Add 5 bonus marks only to scores below 50, keep others unchanged
adjusted_scores = np.where(scores < 50, scores + 5, scores)
print("Adjusted Scores (+5 if < 50):", adjusted_scores)

# 3. Using np.where to find indices of matching elements
honor_roll_idx = np.where(scores >= 85)[0]
print("\nIndices of scores >= 85:", honor_roll_idx)
print("Honor roll scores:", scores[honor_roll_idx])
```

### 2. Handling Missing Values (`NaN`): `np.isnan()`, `np.nanmean()`, `np.nan_to_num()`
Real-world datasets frequently contain missing or corrupted measurements represented as `np.nan` (Not a Number, of `float` type).
- **Problem**: Standard aggregation functions (`np.mean()`, `np.sum()`, `np.std()`) fail and return `nan` when even one `NaN` is present.
- **Detection**: `np.isnan(arr)` returns a boolean mask indicating missing values.
- **Safe Aggregation**: Functions prefixed with `nan` (`np.nanmean`, `np.nansum`, `np.nanstd`, `np.nanmedian`) automatically skip `NaN` values.
- **Imputation**: `np.nan_to_num(arr, nan=replacement_value)` fills missing values with a designated number (such as 0 or the column mean).


```python
# Sample data with missing values (np.nan is a float)
raw_data = np.array([85.0, 92.0, np.nan, 78.0, np.nan, 88.0])
print("Raw data with missing values:", raw_data)

# Problem: Standard aggregation fails when NaNs exist
print("\nStandard mean (returns NaN):", np.mean(raw_data))
print("Standard sum (returns NaN):", np.sum(raw_data))

# Solution 1: Detecting and counting missing values
nan_mask = np.isnan(raw_data)
print("\nNaN mask:", nan_mask)
print("Count of missing values:", np.sum(nan_mask))

# Solution 2: Filtering out missing values using boolean negation (~)
clean_data = raw_data[~nan_mask]
print("Filtered data (without NaNs):", clean_data)

# Solution 3: Using NaN-safe statistical functions
safe_mean = np.nanmean(raw_data)
safe_std = np.nanstd(raw_data)
print(f"\nNaN-safe Mean (np.nanmean): {safe_mean:.2f}")
print(f"NaN-safe Std (np.nanstd): {safe_std:.2f}")

# Solution 4: Imputing/Replacing NaNs using np.nan_to_num
imputed_zero = np.nan_to_num(raw_data, nan=0.0)
imputed_mean = np.nan_to_num(raw_data, nan=safe_mean)
print("Imputed with 0.0:", imputed_zero)
print("Imputed with column mean:", np.round(imputed_mean, 2))
```

### 3. Frequency Analysis & Value Counts: `np.unique()`
In Exploratory Data Analysis (EDA), inspecting the distribution of distinct categories is fundamental.
- `np.unique(arr)` returns sorted unique values.
- `return_counts=True` returns the frequency count for each unique value (equivalent to Pandas `Series.value_counts()`).
- `axis=0` on a 2D matrix enables deduplication of table records.


```python
# Customer feedback ratings (scale 1 to 5)
ratings = np.array([5, 4, 3, 5, 2, 4, 5, 3, 4, 5, 1, 5, 4, 2])
print("Ratings collected:", ratings)

# 1. Unique values and their frequencies (value counts)
unique_ratings, counts = np.unique(ratings, return_counts=True)
print("\nRating Frequency Distribution:")
for val, count in zip(unique_ratings, counts):
    pct = (count / len(ratings)) * 100
    print(f"Rating {val}: {count} times ({pct:.1f}%)")

# 2. Finding unique rows in a 2D matrix (deduplicating tabular records)
transactions = np.array([
    [101, 500],
    [102, 300],
    [101, 500],  # Duplicate record
    [103, 750],
    [102, 300]   # Duplicate record
])
unique_transactions = np.unique(transactions, axis=0)

print("\nOriginal transactions (with duplicates):\n", transactions)
print("Deduplicated unique transactions:\n", unique_transactions)
```

### 4. Sorting & Ranking Multi-dimensional Data: `np.argsort()`
In tabular data analysis, sorting a single column independently breaks the relationship across features!
- `np.sort(arr)` sorts the values directly.
- `np.argsort(arr)` returns the **indices that would sort the array**.
- Using `data[np.argsort(data[:, col_idx])]` sorts the entire dataset by that specific column while keeping all row records intact.
- Append `[::-1]` for descending order (e.g. top performers, highest sales).


```python
# Student dataset: [Student_ID, Math_Marks, Science_Marks]
students = np.array([
    [1, 78, 92],
    [2, 95, 81],
    [3, 62, 70],
    [4, 88, 95],
    [5, 91, 65]
])

print("Original Students Data [ID, Math, Science]:\n", students)

# 1. Sort entire table by Math score (column index 1) - Ascending
math_sort_indices = np.argsort(students[:, 1])
print("\nIndices sorting by Math:", math_sort_indices)
print("Table sorted by Math (ascending):\n", students[math_sort_indices])

# 2. Sort entire table by Science score (column index 2) - Descending
sci_desc_indices = np.argsort(students[:, 2])[::-1]
print("\nIndices sorting by Science (descending):", sci_desc_indices)
print("Table sorted by Science (highest to lowest):\n", students[sci_desc_indices])

# 3. Top 2 performers in Science
top_2_science = students[sci_desc_indices[:2]]
print("\nTop 2 Students in Science:\n", top_2_science)
```

### 5. Outlier Capping & Clamping: `np.clip()`
Extreme values or measurement errors (outliers) can heavily distort statistical summaries like the mean and standard deviation.
- `np.clip(a, a_min, a_max)` restricts all array values to lie strictly within `[a_min, a_max]`.
- Any value `< a_min` is capped to `a_min`; any value `> a_max` is capped to `a_max`.
- In Data Analytics, this is known as **Winsorization**, a standard preprocessing step to mitigate the impact of outliers without dropping rows.


```python
# 1. Clamping exam scores to valid boundaries [0, 100]
raw_scores = np.array([45, 88, -12, 105, 92, 115, 76, -5, 60])
print("Raw scores (with erroneous negatives and >100 entries):\n", raw_scores)

capped_scores = np.clip(raw_scores, a_min=0, a_max=100)
print("\nCapped scores (restricted to [0, 100]):\n", capped_scores)

# 2. Practical Data Analytics: Winsorizing salary outliers using percentiles
salaries = np.array([25000, 32000, 28000, 150000, 30000, 27000, 31000, 350000])
p5 = np.percentile(salaries, 5)
p95 = np.percentile(salaries, 95)

# Cap extreme salaries to the 5th and 95th percentiles
winsorized_salaries = np.clip(salaries, p5, p95)

print("\nOriginal Salaries:", salaries)
print(f"5th percentile threshold: {p5:.0f}")
print(f"95th percentile threshold: {p95:.0f}")
print("Winsorized Salaries (outliers capped):", winsorized_salaries.astype(int))
```

## Applied Practice: Data Analytics Essentials in Action
Hands-on exercises applying `np.where()`, missing value imputation (`np.nan`), categorical frequency analysis (`np.unique`), multi-dimensional sorting (`np.argsort`), and outlier capping (`np.clip`) to real student analytics data.


```python
# Initializing student dataset for applied practice
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 90, 78],   # Student 1
    [19, 97, 88],   # Student 2
    [17, 81, 95],   # Student 3
    [18, 70,  0],   # Student 4
    [20, 95, 85]    # Student 5
])

print("Student Dataset Initialized:")
print("Columns: [Age, Math Marks, Science Marks]")
print(data)
```

### Exercise 1: Categorical Flagging with `np.where()`
**Goal**: Create a new performance flag for each student:
- If Math score >= 85, mark as `"High Performer"`, otherwise `"Standard"`.
- Display each student's age, score, and resulting category.


```python
# Create binary performance flag using np.where
performance_flag = np.where(data[:, 1] >= 85, "High Performer", "Standard")
print("Performance Flag (Math >= 85):", performance_flag)

# Display student results with category
for i in range(len(data)):
    print(f"Student {i+1} (Age {data[i,0]}): Math {data[i,1]} -> {performance_flag[i]}")
```

### Exercise 2: Handling Missing Exam Scores (`NaN`)
**Goal**: In real-world data, students miss exams due to illness. 
- Simulate missing scores by converting the array to float and inserting `np.nan`.
- Show how standard `np.mean()` fails when `NaN` is present.
- Calculate the class average ignoring absent students using `np.nanmean()`.
- Impute missing scores with the class average using `np.nan_to_num()`.


```python
# Create a float copy to support np.nan
data_with_absentee = data.astype(float).copy()

# Student 4 was absent for Science exam; Student 3 had unrecorded Math score
data_with_absentee[3, 2] = np.nan
data_with_absentee[2, 1] = np.nan

print("Dataset with Missing Exam Scores (NaN):\n", data_with_absentee)
print("\nMissing values count in dataset:", np.isnan(data_with_absentee).sum())

# Standard mean fails:
print("Standard mean of Science marks (returns NaN):", np.mean(data_with_absentee[:, 2]))

# NaN-safe average:
safe_sci_mean = np.nanmean(data_with_absentee[:, 2])
safe_math_mean = np.nanmean(data_with_absentee[:, 1])
print(f"NaN-safe Mean Science Score: {safe_sci_mean:.2f}")
print(f"NaN-safe Mean Math Score: {safe_math_mean:.2f}")

# Impute missing marks with respective subject average
data_with_absentee[:, 1] = np.nan_to_num(data_with_absentee[:, 1], nan=safe_math_mean)
data_with_absentee[:, 2] = np.nan_to_num(data_with_absentee[:, 2], nan=safe_sci_mean)

print("\nImputed Dataset (Cleaned & ready for downstream analysis):\n", np.round(data_with_absentee, 1))
```

### Exercise 3: Frequency Distribution & Value Counts with `np.unique()`
**Goal**: Analyze demographic and performance distribution.
- Find unique ages and their frequency distribution (`return_counts=True`).
- Calculate percentage share of each age group.


```python
# Analyzing student ages (Column 0)
ages = data[:, 0]
unique_ages, age_counts = np.unique(ages, return_counts=True)

print("Demographic Age Breakdown:")
print(f"{'Age':<6} {'Student Count':<15} {'Percentage':<10}")
print("-" * 35)
for age, count in zip(unique_ages, age_counts):
    pct = (count / len(ages)) * 100
    print(f"{age:<6} {count:<15} {pct:.1f}%")
```

### Exercise 4: Multi-Column Sorting & Ranking with `np.argsort()`
**Goal**: In analytics, we frequently need a "Leaderboard". Sort the entire student table based on total marks in descending order without misaligning student records across columns.


```python
# Calculate total marks (Math + Science)
total_scores = data[:, 1] + data[:, 2]

# np.argsort returns indices in ascending order; [::-1] reverses to descending
rank_order = np.argsort(total_scores)[::-1]

print("Student Leaderboard (Ranked by Total Marks):")
print(f"{'Rank':<6} {'Age':<6} {'Math':<6} {'Science':<8} {'Total':<6}")
print("-" * 35)
for rank, idx in enumerate(rank_order, 1):
    print(f"{rank:<6} {data[idx,0]:<6} {data[idx,1]:<6} {data[idx,2]:<8} {total_scores[idx]:<6}")

top_student_idx = rank_order[0]
print(f"\nTop performing student index: {top_student_idx} with {total_scores[top_student_idx]} total points.")
```

### Exercise 5: Outlier Management with `np.clip()`
**Goal**: Prevent data corruption.
- Suppose an automated grading script introduced an erroneous score of 120 (bonus overflow) and -10 (recording glitch).
- Cap all scores strictly to valid exam bounds [0, 100] using `np.clip()`.


```python
# Simulated raw scores with recording errors:
glitched_scores = np.array([120, 88, -10, 75, 95])
print("Glitched Exam Scores (with out-of-bounds errors):", glitched_scores)

# Use np.clip to restrict scores strictly between 0 and 100
cleaned_scores = np.clip(glitched_scores, a_min=0, a_max=100)
print("Sanitized Exam Scores (Capped [0, 100]):", cleaned_scores)
```
