# NumPy Array Indexing and Slicing
This notebook covers how to access and extract elements from NumPy arrays using various indexing and slicing techniques.

```python
# Import NumPy - essential for all array operations
import numpy as np
```

## NumPy Indexing and Slicing of Vectors (1D Arrays)
Understanding how to access elements in one-dimensional arrays

```python
# Creating a 1D array with numbers from 11 to 20
# This will be our test array for demonstrating indexing concepts
# Index positions: 0=11, 1=12, 2=13, ..., 9=20
arr = np.arange(11,21)
arr
```

```python
# Single element indexing - accessing element at index 6
# Remember: Python uses 0-based indexing
# Index 6 corresponds to the 7th element, which is 17
arr[6]
```

```python
# Negative indexing - accessing elements from the end
# -1 is the last element, -2 is second to last, etc.
print("Last element:", arr[-1])      # 20
print("Second to last:", arr[-2])    # 19
print("Third to last:", arr[-3])     # 18
```

```python
# Slicing: extracting a portion of the array
# Syntax: arr[start:stop] - includes start, excludes stop
# arr[1:5] gets elements at indices 1, 2, 3, 4 (not 5)
# Results in elements: 12, 13, 14, 15
arr[1:5]
```

```python
# Slicing from beginning: omitting start index defaults to 0
# arr[:5] is equivalent to arr[0:5]
# Gets first 5 elements (indices 0 through 4)
arr[:5]
```

```python
# Slicing entire array: arr[:] creates a copy of the whole array
# Useful for creating copies or when you want to be explicit
# Returns all elements from start to end
arr[:]
```

```python
# Slicing to end: omitting stop index goes to the end
# arr[3:] is equivalent to arr[3:len(arr)]
# Gets all elements from index 3 to the end
arr[3:]
```

```python
# Alternative syntax with double colon
# arr[3::] is the same as arr[3:] - both go from index 3 to end
# The double colon allows for step specification (shown next)
arr[3::]
```

```python
# Slicing with step: arr[start:stop:step]
# arr[3::2] starts at index 3, goes to end, takes every 2nd element
# Results: index 3(14), index 5(16), index 7(18), index 9(20)
# Useful for sampling data or extracting patterns
arr_slice = arr[3::2]
arr_slice
```

```python
# Advanced slicing examples

# Reverse the array
print("Reversed array:", arr[::-1])

# Every second element
print("Every 2nd element:", arr[::2])

# Every third element starting from index 1
print("Every 3rd from index 1:", arr[1::3])

# Reverse slice with step
print("Reverse every 2nd:", arr[::-2])

# Complex slicing
print("Middle portion reversed:", arr[2:8][::-1])
```

## NumPy Indexing and Slicing of Matrices (2D Arrays)
Understanding how to access elements, rows, columns, and sub-matrices in two-dimensional arrays

```python
# Creating a 2D array (matrix) with 6 rows and 5 columns
# Using arange(1,31) to create numbers 1-30, then reshape to 6x5 matrix
# This creates a matrix where each row contains 5 consecutive numbers
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
# Accessing entire first row (index 0)
# Returns a 1D array containing all elements from the first row
# Row indexing: arr[row_index] gives you the entire row
arr[0]
```

```python
# Accessing entire last row (index 5)
# Since we have 6 rows (0-5), index 5 is the last row
# Contains the highest values: 26, 27, 28, 29, 30
arr[5]
```

```python
# Accessing specific element using double bracket notation
# arr[0][3] means: first get row 0, then get element at index 3
# This is equivalent to arr[0,3] but less efficient
# Gets element at row 0, column 3 = 4
arr[0][3]
```

```python
# Accessing specific element using comma notation (preferred method)
# arr[row, column] - more efficient than double brackets
# arr[5,4] gets element at row 5, column 4 = 30 (bottom-right corner)
arr[5,4]
```

```python
# 2D slicing: extracting a sub-matrix
# arr[row_start:row_stop, col_start:col_stop]
# arr[0:2, 1:3] gets rows 0-1 and columns 1-2
# Results in a 2x2 matrix with elements: [[2,3], [7,8]]
slice_arr = arr[0:2,1:3]
slice_arr
```

```python
# Slicing from specific point to end
# arr[3:, 3:] gets all rows from index 3 onwards and all columns from index 3 onwards
# This extracts the bottom-right corner of the matrix
# Results in a 3x2 matrix with the last 2 columns of the last 3 rows
slice_arr = arr[3:,3:]
slice_arr
```

```python
# Extracting entire column using colon notation
# arr[:, column_index] gets all rows for a specific column
# arr[:,3] gets column 3 (4th column) from all rows
# Results in 1D array: [4, 9, 14, 19, 24, 29]
arr[:,3]
```

```python
# Advanced matrix indexing examples

# Extract multiple specific rows
print("Rows 0, 2, 4:")
print(arr[[0, 2, 4]])

# Extract multiple specific columns
print("\nColumns 1, 3:")
print(arr[:, [1, 3]])

# Extract specific rows and columns combination
print("\nRows 1,3 and Columns 0,2,4:")
print(arr[np.ix_([1, 3], [0, 2, 4])])

# Negative indexing in 2D
print("\nLast row, last column:", arr[-1, -1])
print("Last 2 rows, first 3 columns:")
print(arr[-2:, :3])

# Step slicing in 2D
print("\nEvery other row and column:")
print(arr[::2, ::2])
```

## Boolean Indexing
Using boolean conditions to filter and select elements from arrays

```python
# Creating a fresh array for boolean indexing demonstration
# Using numbers 11-20 to show filtering techniques
arr = np.arange(11,21)
arr
```

```python
# Creating a boolean mask to find even numbers
# arr % 2 == 0 checks each element: if divisible by 2, returns True, else False
# This creates a boolean array with same shape as original array
# True for even numbers (12, 14, 16, 18, 20), False for odd numbers
bool_index = arr%2 == 0
bool_index
```

```python
# Applying boolean mask to filter the array
# arr[bool_index] returns only elements where bool_index is True
# This extracts all even numbers from the original array
# Result: only the even numbers [12, 14, 16, 18, 20]
arr = arr[bool_index]
arr
```

```python
# Advanced boolean indexing examples

# Reset array for more examples
arr = np.arange(1, 21)
print("Original array:", arr)

# Multiple conditions with logical operators
# Find numbers greater than 10 AND less than 15
condition = (arr > 10) & (arr < 15)
print("Numbers between 10 and 15:", arr[condition])

# Find numbers less than 5 OR greater than 15
condition = (arr < 5) | (arr > 15)
print("Numbers < 5 or > 15:", arr[condition])

# Find numbers NOT equal to specific values
condition = ~np.isin(arr, [5, 10, 15])
print("Numbers excluding 5, 10, 15:", arr[condition])

# Boolean indexing with 2D arrays
matrix = np.arange(1, 25).reshape(4, 6)
print("\nOriginal matrix:")
print(matrix)

# Find all elements greater than 15
print("\nElements > 15:", matrix[matrix > 15])

# Replace elements based on condition
matrix_copy = matrix.copy()
matrix_copy[matrix_copy < 10] = 0  # Replace values < 10 with 0
print("\nMatrix with values < 10 replaced by 0:")
print(matrix_copy)
```

## Fancy Indexing
Using arrays of indices to select multiple elements at once

```python
# Fancy indexing with 1D arrays
arr = np.arange(10, 50, 5)  # [10, 15, 20, 25, 30, 35, 40, 45]
print("Original array:", arr)

# Select elements at specific indices
indices = [1, 3, 5, 7]  # Select 2nd, 4th, 6th, 8th elements
print("Elements at indices [1,3,5,7]:", arr[indices])

# Fancy indexing with 2D arrays
matrix = np.arange(1, 25).reshape(4, 6)
print("\nOriginal matrix:")
print(matrix)

# Select specific rows
row_indices = [0, 2, 3]
print("\nRows 0, 2, 3:")
print(matrix[row_indices])

# Select specific elements using row and column indices
row_idx = [0, 1, 2, 3]
col_idx = [1, 2, 3, 4]
print("\nDiagonal-like selection:", matrix[row_idx, col_idx])

# Advanced fancy indexing - select submatrix
rows = [1, 3]
cols = [0, 2, 4]
print("\nSubmatrix (rows 1,3 and cols 0,2,4):")
print(matrix[np.ix_(rows, cols)])
```

## Summary of NumPy Indexing Techniques

1. **Basic Indexing**: `arr[index]` for single elements
2. **Slicing**: `arr[start:stop:step]` for ranges
3. **Multi-dimensional**: `arr[row, col]` for 2D arrays
4. **Boolean Indexing**: `arr[condition]` for filtering
5. **Fancy Indexing**: `arr[index_array]` for multiple specific elements
6. **Negative Indexing**: `arr[-1]` for counting from the end

These techniques are fundamental for data manipulation and analysis in NumPy!
